To enable login in the streamlit app while providing for testing or portability of CRUD functions, engine creation should be handled by its own module and login errors need to be tested. This notebook provides a space for that.

```python
from sqlalchemy import create_engine, text
from getpass import getpass

schema = "sample_library"
host = "127.0.0.1"
user = "root"
password = getpass("Password")
port = 3306

connection_string = f'mysql+pymysql://{user}:{password}@{host}:{port}/{schema}'

engine = create_engine(connection_string)
try:
    engine.connect()
    print("login succeeded")
except RuntimeError:
    print("login failed")
```

```python
import pandas as pd

pd.read_sql("friends", con=engine)
```
